```python
import os
import glob
import pandas as pd
from datetime import date

# Mount Google Drive (so you can read/write Excel files)
from google.colab import drive
drive.mount('/content/drive')

# Path to your folder in Drive
folder_path = "/content/drive/MyDrive/Excel_Automation_Project"  # <-- change this
snapshot_path = "/content/drive/MyDrive/Excel_Automation_Project/snapshot" 

# Output file path
output_file = os.path.join(folder_path, "combined_output.xlsx")

# 🔹 Delete previous output file if it exists
if os.path.exists(output_file):
    os.remove(output_file)
    print("🗑️ Previous combined_output.xlsx deleted")

# Find all Excel files (skip the combined output itself)
excel_files = [
    f for f in glob.glob(os.path.join(folder_path, "*.xlsx"))
    if os.path.basename(f) != "combined_output.xlsx"
]

# Show list of files found
print("📂 Excel files to combine:")
for f in excel_files:
    print("   -", os.path.basename(f))

dfs = []
for file in excel_files:
    df = pd.read_excel(file)
    df["SourceFile"] = os.path.basename(file)
    dfs.append(df)

combined_df = pd.concat(dfs, ignore_index=True)

# Save combined file back into Drive
output_file = os.path.join(folder_path, "combined_output.xlsx")
combined_df.to_excel(output_file, index=False)

# Save daily snapshot with today's date
snapshot_file = os.path.join(snapshot_path, f"combined_output_{date.today()}.xlsx")
combined_df.to_excel(snapshot_file, index=False)

print(f"\n✅ All Excel files combined into {output_file}")
print(f"📸 Daily snapshot saved: {snapshot_file}")
```
